```python
# build dataset
import matplotlib.pyplot as plt
import matplotlib.image as mpimg 
import os
import numpy as np

msj_path = base_path + '/RoadPoles-MSJ/'
msj_folders = ['20250319_083408', '20250319_083810', '20250319_143033', 
               '20250319_150050', '20250319_083536','20250319_142928', '20250319_143517']

for j in range(len(msj_folders)):
    folder = msj_folders[j]
    train_img_folder_path = msj_path + folder
    train_frames = []
    for fname in os.listdir(train_img_folder_path):
        if not fname[-4:] == 'json':
            train_frames.append(fname)
    
    nframes = len(train_frames)
    for i in range(0,nframes,nframes//5):
        frame = train_frames[i]
        print(f'folder {j}, {folder=}')
        print(f'img {i} / {nframes}, {frame=}')
        img_path = train_img_folder_path + '/' + frame
        img = mpimg.imread(img_path)   
        img = np.rot90(img, k=-1)
    
        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.axis('off')
        plt.show()
        
    
    
```
